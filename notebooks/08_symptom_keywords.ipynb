# Keyword symptom extraction

Extract symptoms directly from `docs/keywords/keywords_master.csv` using the
Python standard library. No NLP model, training, or annotated dataset is required.
The disease services use this same extractor.


```python
from pathlib import Path
import json
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "healthph_symptoms").is_dir():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from healthph_symptoms.keywords import KeywordSymptomExtractor

keyword_extractor = KeywordSymptomExtractor(ROOT / "docs/keywords/keywords_master.csv")
text = "May ubo at lagnat ako. Walang sipon."
print(json.dumps(keyword_extractor.extract(text), ensure_ascii=False, indent=2))
```
